## 3 Character strings

Character strings (strings) are used to store textual information.

A *string* is defined by putting the information between single quotes
`'` or double quotes `"`.

### 3.1 Definition

```python
a = 'a string defined with single quotes'
a
```

```python
b = "a string defined with double quotes"
b
```

```python
# To include an apostrophe inside the string
"j'inclus l'apostrophe"
```

```python
# To include double quotes inside the string
'les "guillemets" sont là'
```

### 3.2 Useful methods

Here is an overview of some useful methods. Many others exist (see
[official
documentation](https://docs.python.org/3/library/stdtypes.html#string-methods))

```python
# String length (number of characters)
len("J'ai 18 caractères")
```

```python
# Concatenation
"I am" + "your father"
" ".join(["I am", "your father"])
```

```python
# Concatenation with a number
year = 2
"je suis en " + year + "ème année"
```

Problem: the number must first be converted into a string. The **str**
method does the job.

```python
# Concatenation with a number - after converting the number to a string
year = 2
"je suis en " + str(year) + "ème année"
```

```python
# Repetition
"hop " * 5
```

```python
# Convert to uppercase
"C'est OK".upper()
```

```python
# Convert to lowercase
"C'est OK".lower()
```

```python
# Count the number of occurrences
"Mangez cinq fruits et légumes par jour".count("a")
```

```python
# Create a list of words
"Mangez cinq fruits et légumes par jour".split()
```

```python
# Split words according to a specific character
"un-deux-trois-soleil".split("-")
```

```python
# Use strings as templates
"mon numéro est : {}".format("06 12 34 56 78")
```

```python
# Starts with ?
"vélo".startswith("vé")
```

```python
# Ends with ?
"vélo".endswith("lo")
```

### 3.3 Extracting substrings

A string is considered in Python as a list of characters. It is
therefore possible to extract different elements from this list.

```python
# First element
"c'est de toute beauté"[0]
```

```python
# Second element
"c'est de toute beauté"[1]
```

```python
# Last element
"c'est de toute beauté"[-1]
```

```python
# From a specific character
"c'est de toute beauté"[6:]
```

```python
# Up to a specific character
"c'est de toute beauté"[:8]
```

```python
# Extract a substring
"c'est de toute beauté"[9:14]
```

```python
# Extract every 2 characters, starting from the 4th position
"c'est de toute beauté"[4::2]
```

```python
# Reverse a string
"c'est de toute beauté"[::-1]
```

#### Special characters

To insert special characters within a string, you must use the escape
character `\`.

| Character        | Description      |
|------------------|------------------|
| <br>             | Escape character |
| '                | Apostrophe       |
| "                | Double quotes    |
|                  | New line         |
|  Horizontal tab  |                  |
|  Carriage return |                  |

```python
c = "une chaîne\nsur 2 lignes"
print(c)
```

```python
# Define a string over multiple lines
d = """une autre \"possibilité\" pour
avoir une chaîne sur 2 lignes"""
print(d)
```

### 4.1 Comparison operators

| Operator | Meaning                  |
|----------|--------------------------|
| ==       | Equal to                 |
| !=       | Not equal to             |
| \<       | Strictly less than       |
| \>       | Strictly greater than    |
| \<=      | Less than or equal to    |
| \>=      | Greater than or equal to |

```python
8 > 5
```

```python
1+1 == 2
```

```python
[1, 2, 3] == [1, 2, 3]
```

```python
"girafe" != "gnou"
```

```python
# Chained operators
1 < 2 == 2 >= 1 != 2
```
